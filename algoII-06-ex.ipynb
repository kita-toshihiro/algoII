### 課題 1：オイラー法と 4次ルンゲ＝クッタ法の手計算追跡と空欄穴埋め

講義本編「2. 基本実装と動作検証」で扱った次の 1階常微分方程式の初期値問題を考えます：

$$
\frac{dy}{dt} = f(t, y) = y - t^2 + 1, \quad y(0) = 0.5 \quad (0 \le t \le 2)
$$

この方程式の数学的な厳密解（真解）は以下で与えられます：
$$
y_{\text{true}}(t) = (t + 1)^2 - 0.5 e^t
$$

手計算が最もスムーズに行えるよう、歩幅（刻み幅）として **$h = 1.0$** を採用します。  
初期値 $(t_0, y_0) = (0.0, 0.5)$ から出発して、区間 $[0, 2]$ を次の 2 ステップで解く過程を追跡します：
- **第 1 ステップ**: $t_0 = 0.0 \to t_1 = 1.0$
- **第 2 ステップ**: $t_1 = 1.0 \to t_2 = 2.0$

---
#### ■ Part 1：オイラー法（Euler's Method）による計算過程

##### 【第 1 ステップ ($t_0 = 0.0 \to t_1 = 1.0$)】
スタート地点 $(t_0, y_0) = (0.0, 0.5)$ において、オイラー法を用いて 1 ステップ進める計算を行います。

1. **スタート地点での傾き**:
   $$ f(t_0, y_0) = \text{[ 空欄 A ]} $$

2. **第 1 ステップの更新値**:
   オイラー法の考え方（足元の傾きに沿って歩幅 $h = 1.0$ だけ直線で進む）に基づいて次の一歩を計算すると：
   $$ y_1 = \text{[ 空欄 B ]} $$

##### 【第 2 ステップ ($t_1 = 1.0 \to t_2 = 2.0$)】
第 1 ステップで求まった $(t_1, y_1) = (1.0, 2.0)$ を現在地として、同様に第 2 ステップを進めます：
- 現在地での傾き：$f(t_1, y_1) = 2.0 - 1.0^2 + 1 = 2.0$
- 第 2 ステップの更新値：$y_2 = y_1 + h \cdot f(t_1, y_1) = 2.0 + 1.0 \times 2.0 = 4.0$

---
#### ■ Part 2：4次ルンゲ＝クッタ法（RK4）による計算過程

##### 【第 1 ステップ ($t_0 = 0.0 \to t_1 = 1.0$)】
RK4 では、区間 $[0.0, 1.0]$ の中で 4 つの傾き $k_1, k_2, k_3, k_4$ をサンプリング（下見）し、加重平均して次の一歩を決定します。

1. **$k_1$（現在地での傾き）**:
   スタート地点 $(t_0, y_0) = (0.0, 0.5)$ における傾きです（上記オイラー法と同じ）：
   $$ k_1 = \text{[ 空欄 A ]} $$

2. **$k_2$（中点 $t = 0.5$ での下見 ①）**:  
   傾き $k_1$ を頼りに半歩（$h/2 = 0.5$）進んだ中点での傾きを調べます：
   - 中点における予測位置：
     $$ y = y_0 + \frac{h}{2} k_1 = \text{[ 空欄 C ]} $$
   - その中点における傾き：
     $$ k_2 = f(0.5, \text{[ 空欄 C ]}) = \text{[ 空欄 D ]} $$

3. **$k_3$（中点 $t = 0.5$ での下見 ②）**:  
   より精度の上がった傾き $k_2$ を頼りに、もう一度同じ中点（$t = 0.5$）での傾きを見直します：
   - その中点における傾き：
     $$ k_3 = \text{[ 空欄 E ]} $$

4. **$k_4$（終点 $t = 1.0$ での下見）**:  
   中点の傾き $k_3$ を頼りに、一気に区間の終点（$t = 1.0$）まで進んだ地点での傾きを調べます：
   - その終点における傾き：
     $$ k_4 = \text{[ 空欄 F ]} $$

5. **第 1 ステップの更新値**:  
   4つの傾き $k_1, k_2, k_3, k_4$ を加重平均ブレンドし、歩幅 $h = 1.0$ 進めた次の一歩 $y_1$ を求めます：
   $$ y_1 = \text{[ 空欄 G ]} $$

##### 【第 2 ステップ ($t_1 = 1.0 \to t_2 = 2.0$)】
第 1 ステップで得られた $(t_1, y_1) = (1.0, 2.625)$ から、同様の手順で 4 つの傾きを計算します：
- $k_1 = f(1.0, 2.625) = 2.625 - 1.0^2 + 1 = 2.625$
- 中点下見 ① ($t = 1.5$): $y = 2.625 + 0.5 \times 2.625 = 3.9375 \implies k_2 = f(1.5, 3.9375) = 3.9375 - 2.25 + 1 = 2.6875$
- 中点下見 ② ($t = 1.5$): $y = 2.625 + 0.5 \times 2.6875 = 3.96875 \implies k_3 = f(1.5, 3.96875) = 3.96875 - 2.25 + 1 = 2.71875$
- 終点下見 ($t = 2.0$): $y = 2.625 + 1.0 \times 2.71875 = 5.34375 \implies k_4 = f(2.0, 5.34375) = 5.34375 - 4.0 + 1 = 2.34375$
- 傾きの加重平均ブレンド:
  $$ k_{\text{blend}} = \frac{k_1 + 2k_2 + 2k_3 + k_4}{6} = \frac{15.78125}{6} \approx 2.630208 $$
- 第 2 ステップの更新値:
  $$ y_2 = y_1 + h \cdot k_{\text{blend}} = 2.625 + 1.0 \times 2.630208 \approx 5.255208 $$

##### 【真解（厳密解）との比較】
- **時刻 $t = 1.0$ (Step 1)**: 真解 $y_{\text{true}}(1.0) = (2.0)^2 - 0.5 e^{1.0} \approx 2.640859$
  - オイラー法誤差: $|2.0000 - 2.640859| \approx 0.6409$
  - RK4 誤差: $|2.6250 - 2.640859| \approx 0.0159$ （RK4 はオイラー法より **約 40 倍正確！**）
- **時刻 $t = 2.0$ (Step 2)**: 真解 $y_{\text{true}}(2.0) = (3.0)^2 - 0.5 e^{2.0} \approx 5.305472$
  - オイラー法誤差: $|4.0000 - 5.305472| \approx 1.3055$
  - RK4 誤差: $|5.2552 - 5.305472| \approx 0.0503$ （RK4 はオイラー法より **約 26 倍正確！**）

### 課題 3：Pythonによる数値検証と可視化

以下のコードセルを実行し、
1. 手計算した $h = 1.0$ の第1ステップ（$t=1.0$）および第2ステップ（$t=2.0$）の数値が、プログラムの実行結果と完全に一致することを確認してください。
2. $h = 1.0$ という「非常に大きな歩幅」であっても、オイラー法が大きく外れていく（$y(2)=4.0$）のに対し、4次ルンゲ＝クッタ法（RK4）が真の解曲線（$y(2) \approx 5.3055$）に驚くほど忠実に追従している様子をグラフで観察してください。

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import math
import japanize_matplotlib

# 1. 微分方程式の右辺 f(t, y) と真の解 y_true(t)
def f(t, y):
    return y - t**2 + 1

def y_true(t):
    return (t + 1)**2 - 0.5 * np.exp(t)

# 2. 計算条件の設定 (h = 1.0, t in [0.0, 2.0])
h = 1.0
t0, t_end = 0.0, 2.0
n_steps = int((t_end - t0) / h)

t_arr = np.linspace(t0, t_end, n_steps + 1)
y_euler = np.zeros(n_steps + 1)
y_rk4 = np.zeros(n_steps + 1)

y_euler[0] = 0.5
y_rk4[0] = 0.5

# 3. オイラー法と RK4 の数値計算ループ
for n in range(n_steps):
    tn = t_arr[n]
    
    # オイラー法の 1 ステップ更新
    y_euler[n + 1] = y_euler[n] + h * f(tn, y_euler[n])
    
    # 4次ルンゲ＝クッタ法 (RK4) の 1 ステップ更新
    k1 = f(tn, y_rk4[n])
    k2 = f(tn + 0.5 * h, y_rk4[n] + 0.5 * h * k1)
    k3 = f(tn + 0.5 * h, y_rk4[n] + 0.5 * h * k2)
    k4 = f(tn + h, y_rk4[n] + h * k3)
    y_rk4[n + 1] = y_rk4[n] + (h / 6.0) * (k1 + 2.0 * k2 + 2.0 * k3 + k4)

# 4. 数値テーブルの出力（手計算との照合用）
print(f"{'Step':>5} | {'時刻 t':>6} | {'Euler解':>10} | {'RK4解':>10} | {'真解':>10} | {'Euler誤差':>12} | {'RK4誤差':>12}")
print("-" * 75)
for n in range(len(t_arr)):
    tn = t_arr[n]
    ye = y_euler[n]
    yr = y_rk4[n]
    yt = y_true(tn)
    print(f"{n:5d} | {tn:6.2f} | {ye:10.4f} | {yr:10.4f} | {yt:10.4f} | {abs(ye - yt):12.4e} | {abs(yr - yt):12.4e}")
print("-" * 75)
print(f"時刻 t=1.0 (Step 1): Euler={y_euler[1]:.4f}, RK4={y_rk4[1]:.6f}, 真解={y_true(1.0):.6f}")
print(f"時刻 t=2.0 (Step 2): Euler={y_euler[2]:.4f}, RK4={y_rk4[2]:.6f}, 真解={y_true(2.0):.6f}")
print(f"時刻 t=2.0 での比較: RK4 はオイラー法より 約 {abs(y_euler[2] - y_true(2.0)) / abs(y_rk4[2] - y_true(2.0)):.0f} 倍 正確！")

# 5. 可視化（解曲線の比較プロット）
fig, ax = plt.subplots(figsize=(10, 6))

t_fine = np.linspace(t0, t_end, 200)
ax.plot(t_fine, y_true(t_fine), 'b-', lw=2.8, label='真の厳密解 $y(t)$')
ax.plot(t_arr, y_euler, 'ro--', lw=2.2, markersize=8, label=f'オイラー法 ($h={h}$)')
ax.plot(t_arr, y_rk4, 'gs-', lw=2.2, markersize=8, label=f'4次ルンゲ＝クッタ法 RK4 ($h={h}$)')

ax.set_title(f'微分方程式の解曲線比較 ($h={h}$): オイラー法 vs 4次ルンゲ＝クッタ法', fontsize=15, pad=12)
ax.set_xlabel('時刻 $t$', fontsize=13)
ax.set_ylabel('状態 $y(t)$', fontsize=13)
ax.tick_params(labelsize=12)
ax.grid(True, linestyle='--', alpha=0.6)
ax.legend(fontsize=12, loc='upper left')

# 手計算の Step 1 と Step 2 に注釈を追加
ax.annotate('Step 1 (t=1.0)\n手計算の第1地点', xy=(1.0, y_rk4[1]), xytext=(0.55, y_rk4[1] + 0.8),
            arrowprops=dict(arrowstyle='->', color='green', lw=1.6), fontsize=11, color='green', weight='bold')
ax.annotate('Step 2 (t=2.0)\n手計算の第2地点', xy=(2.0, y_rk4[2]), xytext=(1.45, y_rk4[2] + 0.3),
            arrowprops=dict(arrowstyle='->', color='green', lw=1.6), fontsize=11, color='green', weight='bold')

plt.tight_layout()
plt.show()
```
